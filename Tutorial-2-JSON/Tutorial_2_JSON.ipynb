```python
students = [
    {
        "first_name": "Yael",
        "last_name": "Wroblevsky",
        "email": "Yael.Wroblevsky@e.braude.ac.il",
        "courses": ["Cloud", "Web"],
        "link": "https://www.google.com/?hl=iw"
    },
    {
        "first_name": "Shiraz",
        "last_name": "Shamai",
        "email": "Shiraz.Shamai@e.braude.ac.il",
        "courses": ["Cloud", "Web"],
        "link": "https://w3.braude.ac.il/"
    }
]
```

```python
import json

file_path = "/content/drive/MyDrive/students.json"

with open(file_path, "w", encoding="utf-8") as file:
    json.dump(students, file, indent=4, ensure_ascii=False)
```

```python
with open(file_path, "r", encoding="utf-8") as file:
    students = json.load(file)

for student in students:
    print(student)
```

```python
import ipywidgets as widgets
from IPython.display import display

student_dropdown = widgets.Dropdown(
    options=[
        student["first_name"] + " " + student["last_name"]
        for student in students
    ],
    description="Student:"
)

display(student_dropdown)
```

```python
first_name_box = widgets.Text(
    description="First Name:",
    disabled=True
)

last_name_box = widgets.Text(
    description="Last Name:",
    disabled=True
)

email_box = widgets.Text(
    description="Email:",
    disabled=True
)

courses_box = widgets.Textarea(
    description="Courses:",
    disabled=True
)

link_box = widgets.Text(
    description="Link:",
    disabled=True
)

display(
    first_name_box,
    last_name_box,
    email_box,
    courses_box,
    link_box
)
```

```python
def show_student(change):
    selected_name = change["new"]

    for student in students:
        full_name = student["first_name"] + " " + student["last_name"]

        if full_name == selected_name:
            first_name_box.value = student["first_name"]
            last_name_box.value = student["last_name"]
            email_box.value = student["email"]
            courses_box.value = ", ".join(student["courses"])
            link_box.value = student["link"]
            break

student_dropdown.observe(show_student, names="value")

show_student({"new": student_dropdown.value})
```

```python
favorite_show_box = widgets.Text(
    description="Favorite Show:",
    placeholder="Enter favorite show"
)

update_button = widgets.Button(
    description="Update",
    button_style="success"
)

message = widgets.Output()

display(favorite_show_box, update_button, message)
```

```python
def update_student(b):
    selected_name = student_dropdown.value
    favorite_show = favorite_show_box.value

    for student in students:
        full_name = student["first_name"] + " " + student["last_name"]

        if full_name == selected_name:
            student["favorite_show"] = favorite_show
            break

    with open(file_path, "w", encoding="utf-8") as file:
        json.dump(students, file, indent=4, ensure_ascii=False)

    with message:
        message.clear_output()
        print("Student updated successfully!")
update_button.on_click(update_student)
```

```python
with open(file_path, "r", encoding="utf-8") as file:
    updated_students = json.load(file)

for student in updated_students:
    print(student["first_name"], student["last_name"])
    print("Favorite Show:", student.get("favorite_show", "Not set"))
    print()
```
