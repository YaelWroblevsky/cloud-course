```python
from firebase import firebase

FBconn = firebase.FirebaseApplication(
    'https://cloudtutorial3-default-rtdb.firebaseio.com/',
    None
)

print("Firebase connection created!")
```

```python
result = FBconn.get('/', None)

print(result)
```

```python
temperature = 25

data_to_upload = {
    'Temp': temperature
}

result = FBconn.post('/myTest1/', data_to_upload)

print(result)
```

```python
records = FBconn.get('/myTest1/', None)

print(records)
```

```python
record_id = list(records.keys())[0]

result = FBconn.put(
    '/myTest1/' + record_id,
    'Temp',
    30
)

print(result)
```

```python
result = FBconn.delete('/myTest1/', record_id)

print(result)
```

```python
def add_word(word):
    # Read the current count of the word from Firebase
    count = FBconn.get('/words/', word)

    # If the word does not exist, start from zero
    if count is None:
        count = 0

    # Increase the word count by one
    count = count + 1

    # Save the updated count in Firebase
    FBconn.put('/words/', word, count)

    print(f"{word}: {count}")
```

```python
add_word("hello")
add_word("hello")
add_word("world")
```

```python
def add_text(text):
    # Split the text into individual words
    words = text.split()

    # Add each word to Firebase
    for word in words:
        add_word(word)
```

```python
add_text("hello world hello")
```

```python
def update_word(word, new_count):
    # Check if the word already exists in Firebase
    count = FBconn.get('/words/', word)

    if count is None:
        print("Word not found")
        return

    # Update the word count
    FBconn.put('/words/', word, new_count)

    print(f"{word} updated to {new_count}")
```

```python
update_word("hello", 10)
```

```python
def delete_word(word):
    # Check if the word exists in Firebase
    count = FBconn.get('/words/', word)

    if count is None:
        print("Word not found")
        return

    # Delete the word from Firebase
    FBconn.delete('/words/', word)

    print(f"{word} deleted successfully")
```

```python
delete_word("world")
```

```python
def display_words():
    # Get all saved words from Firebase
    words = FBconn.get('/words/', None)

    # Check if the database is empty
    if not words:
        print("No words found")
        return

    # Display all words and their counts
    print("Saved words:")

    for word, count in words.items():
        print(f"{word}: {count}")
```

```python
display_words()
```

```python
while True:
    print("\nCommon Words Counter Menu:")
    print("1. Add single word")
    print("2. Add text for analysis")
    print("3. Update word count")
    print("4. Delete word")
    print("5. View all words")
    print("6. Exit")

    choice = input("\nSelect an option (1-6): ")

    if choice == "1":
        word = input("Enter a word: ")
        add_word(word)

    elif choice == "2":
        text = input("Enter text: ")
        add_text(text)

    elif choice == "3":
        word = input("Enter a word to update: ")
        new_count = int(input("Enter new count: "))
        update_word(word, new_count)

    elif choice == "4":
        word = input("Enter a word to delete: ")
        delete_word(word)

    elif choice == "5":
        display_words()

    elif choice == "6":
        print("Goodbye!")
        break

    else:
        print("Invalid option. Please try again.")
```
